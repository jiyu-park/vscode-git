```python
# 현재 위치 K에서 → 2K로 순간이동 (배터리 0)
# 현재 위치 K에서 → K+1로 점프 (배터리 1 소모)
```

```python
def solution(input_int):
    count = 0

    i = input_int
    while(0<input_int):
        a = input_int % 2 == 0
        if(a):
            # 짝수일때 
           input_int = input_int // 2
        else:
            # 홀수, 1 있으면 + 1 카운트
            input_int -= 1
            count += 1
    return count  




print(solution(5))
print(solution(6))
print(solution(5000))
```
